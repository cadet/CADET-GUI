```python
import ipywidgets as W
from IPython.display import display

from cadetgui.widgets.composite import ConfigurationWidget, DataImportWidget, SolutionWidget

PAGE_CSS = """
<style>
body { background: #f8fafc; }
.cadetgui-app {
  display: flex;
  flex-direction: column;
  gap: 20px;
  max-width: 1100px;
  margin: 0 auto;
  padding: 24px;
  font-family: system-ui, -apple-system, sans-serif;
}
.cadetgui-app-header { margin-bottom: 0; }
.cadetgui-app-header h1 { margin: 0 0 4px 0; font-size: 22px; }
.cadetgui-app-header p { margin: 0; color: #6b7280; font-size: 14px; }
.cadetgui-app-columns { gap: 20px; align-items: flex-start; }
.cadetgui-card {
  background: #ffffff;
  border: 1px solid #e5e7eb;
  border-radius: 10px;
  padding: 16px;
}
/* flex:1 1 0 only makes sense for cards splitting width in a *row* (the two
   columns) — on a card outside that row (data_card, a direct column-flex
   child of .cadetgui-app) it sets flex-basis:0 on the *vertical* axis, and
   with no fixed page height to grow into that collapses the card to 0
   height. Scope it to the row context only. */
.cadetgui-app-columns > .cadetgui-card {
  flex: 1 1 0;
  min-width: 0;
}
@media (prefers-color-scheme: dark) {
  body { background: #111827; }
  .cadetgui-app-header p { color: #9ca3af; }
  .cadetgui-card { background: #1f2937; border-color: #374151; }
}
</style>
"""

header = W.HTML(
    "<div class='cadetgui-app-header'>"
    "<h1>CADET-GUI</h1>"
    "<p>Configure a standard chromatography process, run it, and compare against measured data — powered by CADET-Process.</p>"
    "</div>"
)

config = ConfigurationWidget()
solution = SolutionWidget()
solution.bind_to_config(config)

data = DataImportWidget()
solution.bind_to_data(data)

config_card = W.VBox([config.root])
config_card.add_class("cadetgui-card")
solution_card = W.VBox([solution.root])
solution_card.add_class("cadetgui-card")

columns = W.HBox([config_card, solution_card])
columns.add_class("cadetgui-app-columns")

data_card = W.VBox([data.root])
data_card.add_class("cadetgui-card")

app = W.VBox([W.HTML(PAGE_CSS), header, columns, data_card])
app.add_class("cadetgui-app")

display(app)
```
