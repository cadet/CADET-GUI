```python
import ipywidgets as W
from IPython.display import display

from cadetgui.widgets.composite import ConfigurationWidget, SolutionWidget

PAGE_CSS = """
<style>
body { background: #f8fafc; }
.cadetgui-app {
  max-width: 1100px;
  margin: 0 auto;
  padding: 24px;
  font-family: system-ui, -apple-system, sans-serif;
}
.cadetgui-app-header { margin-bottom: 16px; }
.cadetgui-app-header h1 { margin: 0 0 4px 0; font-size: 22px; }
.cadetgui-app-header p { margin: 0; color: #6b7280; font-size: 14px; }
.cadetgui-app-columns { gap: 20px; align-items: flex-start; }
.cadetgui-card {
  background: #ffffff;
  border: 1px solid #e5e7eb;
  border-radius: 10px;
  padding: 16px;
  flex: 1 1 0;
  min-width: 0;
}
@media (prefers-color-scheme: dark) {
  body { background: #111827; }
  .cadetgui-app-header p { color: #9ca3af; }
  .cadetgui-card { background: #1f2937; border-color: #374151; }
}
</style>
"""

header = W.HTML(
    "<div class='cadetgui-app-header'>"
    "<h1>CADET-GUI</h1>"
    "<p>Configure a standard chromatography process and run it — powered by CADET-Process.</p>"
    "</div>"
)

config = ConfigurationWidget()
solution = SolutionWidget()
solution.bind_to_config(config)

config_card = W.VBox([config.root])
config_card.add_class("cadetgui-card")
solution_card = W.VBox([solution.root])
solution_card.add_class("cadetgui-card")

columns = W.HBox([config_card, solution_card])
columns.add_class("cadetgui-app-columns")

app = W.VBox([W.HTML(PAGE_CSS), header, columns])
app.add_class("cadetgui-app")

display(app)
```
