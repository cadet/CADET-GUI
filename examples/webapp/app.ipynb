```python
import ipywidgets as W
from cadetgui.widgets.composite import (
    ConfigurationWidget,
    InstrumentWidget,
    ParameterEstimationWidget,
    SolutionWidget,
)
from IPython.display import display

PAGE_CSS = """
<style>
body { background: #f8fafc; }
.cadetgui-app {
  max-width: 1100px;
  margin: 0 auto;
  padding: 24px;
  font-family: system-ui, -apple-system, sans-serif;
}
.cadetgui-app-header { margin-bottom: 20px; }
.cadetgui-app-header h1 { margin: 0 0 4px 0; font-size: 22px; }
.cadetgui-app-header p { margin: 0; color: #6b7280; font-size: 14px; }
.cadetgui-tab-page { padding: 20px; }
@media (prefers-color-scheme: dark) {
  body { background: #111827; }
  .cadetgui-app-header p { color: #9ca3af; }
}
</style>
"""

header = W.HTML(
    "<div class='cadetgui-app-header'>"
    "<h1>CADET-GUI</h1>"
    "<p>Configure a standard chromatography process, run it, and compare against "
    "measured data — powered by CADET-Process.</p>"
    "</div>"
)

instrument = InstrumentWidget()
config = ConfigurationWidget(instrument=instrument)
solution = SolutionWidget()
solution.bind_to_config(config)

parameter_estimation = ParameterEstimationWidget()
parameter_estimation.bind_to_config(config)


def tab_page(widget_root: W.Widget) -> W.VBox:
    """Wrap a widget root in the shared tab-page padding class."""
    page = W.VBox([widget_root])
    page.add_class("cadetgui-tab-page")
    return page


tabs = W.Tab(
    children=[
        tab_page(instrument.root),
        tab_page(config.root),
        tab_page(solution.root),
        tab_page(parameter_estimation.root),
    ]
)
tabs.set_title(0, "Instrument")
tabs.set_title(1, "Configuration")
tabs.set_title(2, "Solution")
tabs.set_title(3, "Parameter Estimation")

app = W.VBox([W.HTML(PAGE_CSS), header, tabs])
app.add_class("cadetgui-app")

display(app)
```
