# CADET-GUI: build your own sidebar combination

`WorkbenchWidget` (see `examples/workbench.ipynb`) is a fixed convenience:
it always builds the same three widgets (`ConfigurationWidget`,
`SolutionWidget`, `ParameterEstimationWidget`) and its `include=` kwarg only
lets you turn individual ones of *those three* off -- it cannot add a
widget outside that set.

The actual general-purpose building block is `SidebarShell`
(`cadetgui/widgets/_sidebar_shell.py`): it takes any `(label, widget)`
pairs and shows exactly one at a time. This notebook combines
`ConfigurationWidget` + `SolutionWidget` + a standalone `DataImportWidget`
-- a combination `WorkbenchWidget` cannot express, since `DataImportWidget`
isn't one of its three known steps.

```python
from cadetgui.widgets._sidebar_shell import SidebarShell
from cadetgui.widgets.composite import ConfigurationWidget, DataImportWidget, SolutionWidget
```

## 1. Build whichever widgets you want, in any combination

Nothing here is limited to a fixed set of names -- these are the same
widget classes `WorkbenchWidget` uses internally, constructed directly.

```python
config = ConfigurationWidget()
solution = SolutionWidget()
data = DataImportWidget()
```

## 2. Wire whatever bindings the widgets you picked actually support

`SidebarShell` knows nothing about `bind_to_config`/`add_listener` -- that
wiring is domain-specific to the widgets involved, so it happens here,
before the shell is built (same as `WorkbenchWidget.__init__` does
internally). `DataImportWidget` isn't wired to anything in this example --
it's just browsed alongside the simulation, not bound to it.

```python
solution.bind_to_config(config)
```

## 3. Hand the panes to `SidebarShell`

One `(label, widget)` pair per pane, in whatever order you want them to
appear -- no `include=`, no fixed step names, nothing to opt in or out of
beyond simply not listing a widget here.

```python
shell = SidebarShell(
    [
        ("Configuration", config.root),
        ("Simulation", solution.root),
        ("Experimental Data", data.root),
    ]
)
shell.display()
```

As with every widget in `cadetgui.widgets.composite`, nothing is trapped
inside the shell -- `config`, `solution`, and `data` stay plain attributes
you can keep scripting against after the cell above renders.
